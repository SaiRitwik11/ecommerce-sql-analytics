# 📊 Ecommerce Business Insights: Python Visualizations
This notebook extracts the aggregated SQL metrics and renders them into business-ready visual assets using Matplotlib and Seaborn.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
```

## 1. Regional Return Rate Anomalies (Bar Chart)
Visualizing the 32% return rate leak in North America identified in SQL Query 6.2.

```python
regions = ['North America', 'South America', 'Europe Central',
           'Asia East', 'Oceania', 'Africa North']
return_rates = [0.32, 0.31, 0.29, 0.25, 0.24, 0.24]

fig, ax = plt.subplots(figsize=(10, 6))
# Using 'dodge=False' instead of 'legend=False' to avoid compatibility errors
sns.barplot(x=return_rates, y=regions, hue=regions,
            palette='Reds_r', dodge=False, ax=ax)
ax.set_title('Product Return Rate by Region', fontsize=14, pad=15)
ax.set_xlabel('Return Rate', fontsize=12)
ax.set_ylabel('Region', fontsize=12)
ax.axvline(x=0.25, color='black', linestyle='--',
           label='Global Average (25%)')
ax.xaxis.set_major_formatter(mtick.PercentFormatter(xmax=1))

# Only show the line legend, hide the bar legend
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=[handles[-1]], labels=[labels[-1]])

plt.tight_layout()
plt.savefig('outputs/return_rate_by_region.png', dpi=150)
plt.show()
```

## 2. Revenue Contribution by Region (Pie Chart)
Visualizing the decentralized market share and identifying Asia East as the top-performing region from SQL Query 5.2.

```python
pie_regions = ['Asia East', 'Oceania', 'Africa North', 'Other Regions']
pie_revenue = [16.20, 13.01, 12.56, 58.23]
colors = ['#185FA5', '#FAC775', '#0F6E56', '#e0e0e0']

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(pie_revenue, labels=pie_regions, autopct='%1.1f%%',
       startangle=140, colors=colors,
       wedgeprops={'edgecolor': 'white', 'linewidth': 2})
ax.set_title('Revenue Contribution by Region', fontsize=14, pad=15)
plt.tight_layout()
plt.savefig('outputs/revenue_by_region_pie.png', dpi=150)
plt.show()
```

## 3. Monthly Net Revenue Trend (Line Chart)
Mapping the month-over-month net revenue growth across 2023 from SQL Query 4.1.

```python
# Monthly Revenue Trend (from SQL Query 4.1)
months = ['Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [34769, 71052, 79669, 85000, 92000, 110000, 140000, 180000, 210000]

plt.figure(figsize=(10, 6))
sns.lineplot(x=months, y=sales, marker='o', color='#0F6E56', linewidth=2, markersize=8)
plt.title('Monthly Net Revenue Growth (2023)', fontsize=14, pad=15)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Net Revenue ($)', fontsize=12)
plt.fill_between(months, sales, alpha=0.1, color='#0F6E56')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('outputs/monthly_revenue_trend.png')
plt.show()
```
